## 2) Simuler un user qui a consulté quelques items

On prend 3 ids au hasard et on demande des recommandations.

````python
import joblib, random
idx = joblib.load('models/reco_item_index.pkl')
items = idx['items']
ids = items[idx['id_col']].astype(int).tolist()
viewed = random.sample(ids, 3)
viewed
````

````python
from src.reco.reco_service import load_index, recommend_from_viewed
index = load_index('models/reco_item_index.pkl')
recs = recommend_from_viewed(index, viewed_item_ids=viewed, k=10)
recs[:5]
````
